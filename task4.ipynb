```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel(r"C:\Users\ANAMIKA\Downloads\d4bdb23b-4858-482e-9d51-d9fa09faeec4.xlsx")
df
```

```python
plt.plot(df["Month"],df["Sales"],marker="o",color="green")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.bar(df["Month"],df["Profit"],color="black")
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

```python
plt.scatter(df["Marketing_Spend"],df["Sales"])
plt.title("Marketing Spend vs Sales")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.hist(df['Sales'], bins=6)
plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))


ax[0, 0].plot(df['Month'], df['Sales'], marker='o',color="green")
ax[0, 0].set_title('Monthly Sales')
ax[0, 0].set_xlabel('Month')
ax[0, 0].set_ylabel('Sales')
ax[0, 0].grid(True)


ax[0, 1].bar(df['Month'], df['Profit'],color="black")
ax[0, 1].set_title('Monthly Profit')
ax[0, 1].set_xlabel('Month')
ax[0, 1].set_ylabel('Profit')


ax[1, 0].scatter(df['Marketing_Spend'], df['Sales'],color="purple")
ax[1, 0].set_title('Marketing Spend vs Sales')
ax[1, 0].set_xlabel('Marketing Spend')
ax[1, 0].set_ylabel('Sales')
ax[1, 0].grid(True)


ax[1, 1].plot(df['Month'], df['Customers'], marker='o',color="orange")
ax[1, 1].set_title('Monthly Customers')
ax[1, 1].set_xlabel('Month')
ax[1, 1].set_ylabel('Customers')
ax[1, 1].grid(True)

plt.tight_layout()
plt.show()
```
