#DATA VISUALIZATION

#HISTOGRAM

```python
df.hist(figsize=(12,10))
plt.tight_layout()
plt.show()
```

#PIECHART

```python
Country= df["Country"].value_counts()

plt.pie(
    Country.values,
    labels=Country.index,
    autopct="%1.1f%%"
)

plt.title("country Distribution")
plt.show()
```

#BARCHART

```python
domain_count = df["Domain"].value_counts()

plt.bar(domain_count.index, domain_count.values)

plt.title("Number of Startups by Domain")
plt.xlabel("Domain")
plt.ylabel("Number of Startups")

plt.xticks(rotation=45)
plt.show()
```

#BARCHART

```python
domain_revenue = df.groupby("Domain")["Revenue_ARR_Millions"].mean()

plt.bar(domain_revenue.index, domain_revenue.values)

plt.title("Average Revenue by Domain")
plt.xlabel("Domain")
plt.ylabel("Average Revenue (Million USD)")

plt.xticks(rotation=45)
plt.show()
```

#SCATTERPLOT

```python

sns.scatterplot(
    x="Total_Funding_USD_Millions",
    y="Valuation_USD_Millions",
    hue="Domain",
    data=df
)

plt.title("Funding vs Valuation by Domain")
plt.xlabel("Total Funding (Million USD)")
plt.ylabel("Valuation (Million USD)")

plt.show()
```

#LINECHART

```python
year_count = df["Founding_Year"].value_counts().sort_index()

plt.plot(
    year_count.index,
    year_count.values,
    marker="o"
)

plt.title("Startups by Founding Year")
plt.xlabel("Founding Year")
plt.ylabel("Number of Startups")

plt.grid()
plt.show()
```

```python
cols = [
    "Total_Funding_USD_Millions",
    "Valuation_USD_Millions",
    "Revenue_ARR_Millions",
    "Monthly_Burn_Rate_Millions",
    "Runway_Months_2024",
    "Peak_Headcount_2023",
    "Layoffs_2024_2025",
    "Current_Headcount_2026"
]

corr = df[cols].corr()

corr
```

#HEATMAP

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

#BOXPLOT

```python
sns.boxplot(
    x="Domain",
    y="Total_Funding_USD_Millions",
    data=df
)

plt.title("Funding Distribution by Domain")
plt.xlabel("Domain")
plt.ylabel("Total Funding")
plt.xticks(rotation=45)
plt.show()
```

#HISTPLOT

```python
sns.histplot( df["Total_Funding_USD_Millions"], bins=30, kde=True )
plt.title("Distribution of Total Funding") 
plt.xlabel("Total Funding (Million USD)") 
plt.ylabel("Number of Startups")
plt.show()
```

#KDEPLOT

```python
sns.kdeplot(df["Total_Funding_USD_Millions"], shade=True)
plt.title("Density Distribution of Total Funding")
plt.xlabel("Total Funding (Million USD)") 
plt.ylabel("Density") 
plt.show()
```

#LINEPLOT

sns.lineplot(x="Founding_Year", y="Total_Funding_USD_Millions", data=df) 
plt.title("Founding Year vs Total Funding")
plt.xlabel("Founding Year")
plt.ylabel("Total Funding (Million USD)")
plt.show()

#COUNTPLOT

```python
sns.countplot(x="Country", data=df) 
plt.title("Number of Startups by Country")
plt.xlabel("Country") 
plt.ylabel("Number of Startups")
plt.xticks(rotation=90) 
plt.show()
```

#VIOLINPLOT

```python
sns.violinplot( x="Country", y="Total_Funding_USD_Millions", data=df, color="purple" ) 
plt.title("Funding Distribution by Country") 
plt.xlabel("Country") 
plt.ylabel("Total Funding (Million USD)") 
plt.xticks(rotation=90) 
plt.show()
```

#PAIRPLOT

```python
sns.pairplot(df, hue="Country")
plt.show()
```

```python
sns.jointplot(
    x="Founding_Year",
    y="Runway_Months_2024",
    data=df,
    kind="hex"
)

plt.show()
```
