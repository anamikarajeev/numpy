```python
import pandas as pd

# -----------------------------
# Method 1: Nested List
# -----------------------------
data = [
    [101, "Inception", "Sci-Fi", 8.8, 2010],
    [102, "Titanic", "Romance", 7.9, 1997],
    [103, "Avatar", "Action", 7.8, 2009]
]

columns = ["Movie_ID", "Movie_Name", "Genre", "Rating", "Release_Year"]

df1 = pd.DataFrame(data, columns=columns)

print("Method 1")
print(df1)


```

```python
data = {
    "Movie_ID":[101,102,103],
    "Movie_Name":["Inception","Titanic","Avatar"],
    "Genre":["Sci-Fi","Romance","Action"],
    "Rating":[8.8,7.9,7.8],
    "Release_Year":[2010,1997,2009]
}

df2 = pd.DataFrame(data)

print("\nMethod 2")
print(df2)
```
